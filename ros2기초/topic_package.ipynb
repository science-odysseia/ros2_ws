### Topic 패키지

#### 1. Publisher 

(py_pubsub/py_pubsub/publisher_member_function.py 참고.)

````python
import rclpy
from rclpy.node import Node
from std_msgs.msg import String

class MinimalPublisher(Node):
    def __init__(self):
        super().__init__('minimal_publisher')
        self.publisher_ = self.create_publisher(String, 'topic', 10)
        timer_period = 0.5
        self.timer = self.create_timer(timer_period, self.timer_callback)
        self.i = 0

    def timer_callback(self):
        msg = String()
        msg.data = 'Hello World: %d' % self.i
        self.publisher_.publish(msg)
        self.get_logger().info('Publishing: "%s"' % msg.data)
        self.i += 1
        
def main(args=None):
    rclpy.init(args=args)
    minimal_publisher = MinimalPublisher()
    rclpy.spin(minimal_publisher)
    minimal_publisher.destroy_node()
    rclpy.shutdown()

if __name__ == '__main__':
    main()
````

#### 2. Subscriber

````python
import rclpy
from rclpy.node import Node
from std_msgs.msg import String

class MinimalSubscriber(Node):
    def __init__(self):
        super().__init__('minimal_subscriber')
        self.subscription = self.create_subscription(String, 'topic', self.listener_callback, 10)

    def listener_callback(self, msg):
        self.get_logger().info('I heard: "%s"' % msg.data)


def main(args=None):
    rclpy.init(args=args)
    minimal_subscriber = MinimalSubscriber()
    rclpy.spin(minimal_subscriber)
    minimal_subscriber.destroy_node()
    rclpy.shutdown()

if __name__ == '__main__':
    main()
````

Publisher 코드와 동일한 부분은 생략한다.

````python
class MinimalSubscriber(Node):
````

`Node` 클래스를 상속받는  Subscriber 노드 클래스 `MinimalSubscriber`를 만든다.

````python
super().__init__('minimal_subscriber')
````

부모 클래스 Node의 __init__을 실행하고, 노드 이름을 'minimal_subscriber'로 정한다.

````python
self.subscription = self.create_subscription(String, 'topic', self.listener_callback, 10)
````

`String` 타입 메시지를 구독하는 Publisher인 `self.subscriber_`를 만든다.

`'topic'`이라는 이름의 토픽 통신을 구독하며,

메세지를 받았을 때마다 `self.listener_callback` 함수를 실행한다.

10은 QoS(Quality of Service)로, 최근 메시지를 최대 10개까지 큐에 저장한다.

````python
def listener_callback(self, msg):
    self.get_logger().info('I heard: "%s"' % msg.data)
````

받은 메시지의 data 값을 터미널 로그로 출력한다.

main함수는 publisher main함수와 구조가 똑같으므로 생략한다.

#### 3. setup.py 내용추가

setup.py의 entry_points 안의 'console_scripts'에 아래와 같이 내용을 추가하여 ros2 명령어를 사용할 수 있게 만든다.

````python
entry_points={
    'console_scripts': [
        'talker = py_pubsub.publisher_member_function:main',         # 추가내용
        'listener = py_pubsub.subscriber_member_function:main',      # 추가내용
    ],
}
````

`talker = py_pubsub.publisher_member_function:main`을 예시로 들면

`talker`: 노드 이름의 별칭. 노드 이름은 아님. 노드의 실제 이름은 `super.__init__('노드이름')` 여기서 정했음.

`py_pubsub.publisher_member_function:main` : `py_pubsub` 패키지의 `publisher_member_function` 프로그램의 main함수를 실행하겠다는 의미.

여기서 확장자 .py, .c 등은 넣지 않는다.

위 내용을 추가하고 빌드를 해야 `ros2 run 패키지이름 노드이름` 명령어 실행이 가능하다.
