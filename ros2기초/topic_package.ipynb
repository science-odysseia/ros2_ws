### Topic 패키지

#### 1. Publisher 

````python
import rclpy
from rclpy.node import Node
from std_msgs.msg import String

class MinimalPublisher(Node):
    def __init__(self):
        super().__init__('minimal_publisher')
        self.publisher_ = self.create_publisher(String, 'topic', 10)
        timer_period = 0.5
        self.timer = self.create_timer(timer_period, self.timer_callback)
        self.i = 0

    def timer_callback(self):
        msg = String()
        msg.data = 'Hello World: %d' % self.i
        self.publisher_.publish(msg)
        self.get_logger().info('Publishing: "%s"' % msg.data)
        self.i += 1
        
def main(args=None):
    rclpy.init(args=args)
    minimal_publisher = MinimalPublisher()
    rclpy.spin(minimal_publisher)
    minimal_publisher.destroy_node()
    rclpy.shutdown()

if __name__ == '__main__':
    main()
````

`Node` 클래스를 상속받는 Publisher 노드 클래스 `MinimalPublisher`를 만든다.

````python
super().__init__('minimal_publisher')
````

부모 클래스 Node의 __init__을 실행하고, 노드 이름을 'minimal_publisher'로 정한다.

`ros2 node list`로 노드 이름을 검색할 때,

`/minimal_publisher`로 노드 이름이 뜨게 등록해주는 작업이다.

````python
self.publisher_ = self.create_publisher(String, 'topic', 10)
````

`String` 타입 메시지를 발행하는 Publisher인 `self.publisher_`를 만든다.

`'topic'`이라는 이름의 토픽 통신을 발행한다.

무전기 통신에서 채널을 서로 맞추지 않으면 통신이 불가하듯이,

채널 이름이 `'topic'`이라고 이해하면 된다.

10은 QoS(Quality of Service)이다.

대기열 크기가 10개라는 뜻으로, Subscriber가 잠깐 느려졌을 때 최근 메시지들을 최대 10개까지 보관할 수 있다는 의미이다.

````python
timer_period = 0.5
self.timer = self.create_timer(timer_period, self.timer_callback)
````

0.5초마다 `self.timer_callback` 함수를 실행하는 타이머를 만든다.

````python
self.i = 0
````

뒤에 나오는 `msg.data = 'Hello World: %d' % self.i`에서 사용할 인덱스 변수 `self.i`의 초기값.

String 클래스로 String 타입(내용 : string data)의 메세지를 받아올 클래스 변수 `msg`를 정의한다.

````python
msg.data = 'Hello World: %d' % self.i
````

`msg`의 내용을 정의한다.

````python
self.publisher_.publish(msg)
````

````python
self.get_logger().info('Publishing: "%s"' % msg.data)
````
