\#\# 참고 : Quaternion

Roll, Pitch, Yaw의 경우 Gimbal Lock가 발생할 수 있다.

예를 들어서 만약 Pitch를 90도로 돌리면 

Roll 축과 Yaw 축이 같은 방향으로 겹쳐지고,

이는 Roll 돌리기와 Yaw 돌리기가 똑같은 동작이 되어버리는데 

이를 Gimbal Lock 문제라고 하며

이를 해결하기 위해 나온 것이 Quaternion이다.

3차원의 모든 회전은 

어떤 축(axis)을 기준으로 어떤 각도(θ)만큼 회전 한 것과 

완전히 동일하게 표현할 수 있다는 명제를 기반으로 만들어 졌다.

$q=(x,y,z,w)=(a_x​sin{\frac{θ}{2}}​, a_y​sin{\frac{θ}{2}}​, a_z​sin{\frac{θ}{2}}​, cos{\frac{θ}{2}}​)$

이 식은 회전축의 방향이 $(a_x, a_y, a_z)$, 회전각도가 $\theta$ 인 식이다.

### URDF 실습

#### 01-myfirst.urdf 실습

설치

```bash
sudo apt install ros-humble-urdf-tutorial
```

실행

```bash
ros2 launch urdf_tutorial display.launch.py model:=urdf/01-myfirst.urdf
```

rviz창과 같이 뜨는 또 하나의 창에서 center를 눌러주고

이후 빨간색 원통이 rviz에 뜨는지 확인한다. 시간이 일부 소요된다.

```bash
cat /opt/ros/humble/share/urdf_tutorial/urdf/01-myfirst.urdf
```

로 urdf 내용을 확인한다.

````python
<?xml version="1.0"?>
<robot name="myfirst">
  <link name="base_link">
    <visual>
      <geometry>
        <cylinder length="0.6" radius="0.2"/>
      </geometry>
    </visual>
  </link>
</robot>
````

`<?xml version="1.0"?>` : xml 문서시작. 버전 1.0 사용.

`<robot name="myfirst">` : URDF 최상위 태그. 로봇 이름은 `myfirst`

`<link name="base_link">` : `base_link` 이름의 로봇 부품(여기선 몸통) 생성

`<visual>` : RViz에서 어떻게 보일지를 정의함.

`<geometry>` : 모양정의.

`<cylinder length="0.6" radius="0.2"/>` : 원기둥. 길이(length) 0.6, 반지름(radius) 0.2.

나머지 부분은 태그 종료.

#### 02-multipleshapes.urdf 실습

실행

```bash
ros2 launch urdf_tutorial display.launch.py model:=urdf/02-multipleshapes.urdf
```

rviz창과 같이 뜨는 또 하나의 창에서 center를 눌러주고

이후 rviz에 모양이 나오는지 확인한다. 시간이 일부 소요된다.

내용확인

```bash
cat /opt/ros/humble/share/urdf_tutorial/urdf/02-multipleshapes.urdf
```

````python
<?xml version="1.0"?>
<robot name="multipleshapes">
  <link name="base_link">
    <visual>
      <geometry>
        <cylinder length="0.6" radius="0.2"/>
      </geometry>
    </visual>
  </link>

  <link name="right_leg">
    <visual>
      <geometry>
        <box size="0.6 0.1 0.2"/>
      </geometry>
    </visual>
  </link>

  <joint name="base_to_right_leg" type="fixed">
    <parent link="base_link"/>
    <child link="right_leg"/>
  </joint>

</robot>
````

이번엔 기존의 `base_link`에 더불어

`right_leg` 링크와 `base_to_right_leg` 조인트가 추가되었다.

`right_leg` 링크는 직육면체(`box`)이며, 

x, y, z의 길이는 각각 0.6 0.1 0.2가 된다.

조인트는 Parent link와 child link를 통해 

어느 링크가 부모이고 어느 링크가 자식임을 선언한다.

즉 `base_to_right_leg` 조인트의 부모 링크는 `base_link`, 즉 원통형 몸체이고

자식 링크는 `right_leg`, 즉 직육면체 링크이며,

`type = fixed`를 통해 고정 조임트임을 선언한다.

#### 03-origins.urdf 실습

실행

```bash
ros2 launch urdf_tutorial display.launch.py model:=urdf/03-origins.urdf
```

rviz창과 같이 뜨는 또 하나의 창에서 center를 눌러주고

이후 rviz에 모양이 나오는지 확인한다. 시간이 일부 소요된다.

내용확인

```bash
cat /opt/ros/humble/share/urdf_tutorial/urdf/03-origins.urdf
```

````python
<?xml version="1.0"?>
<robot name="origins">
  <link name="base_link">
    <visual>
      <geometry>
        <cylinder length="0.6" radius="0.2"/>
      </geometry>
    </visual>
  </link>

  <link name="right_leg">
    <visual>
      <geometry>
        <box size="0.6 0.1 0.2"/>
      </geometry>
      <origin rpy="0 1.57075 0" xyz="0 0 -0.3"/>
    </visual>
  </link>

  <joint name="base_to_right_leg" type="fixed">
    <parent link="base_link"/>
    <child link="right_leg"/>
    <origin xyz="0 -0.22 0.25"/>
  </joint>

</robot>
````

이번엔 `right_leg` 링크와 `base_to_right_leg` 조인트에

`<origin>` 요소가 추가되었다.

`<origin>`은 어디에 소속되어 있는지에 따라

1) Link 안의 origin
2) joint 안의 origin

이렇게 2가지로 나눌 수 있는데,

Link 안에 있는 origin은 링크 자체를 어디에 놓고 어떻게 회전시킬지를 정하며,

joint 안에 있는 origin은 

부모 링크 기준으로 자식 링크를 어디에, 어떻게 붙일지를 정해주는 내용이다.

옵션은 xyz, rpy가 있으며, xyz가 위치, rpy(roll, pitch, yaw)가 회전 정보이다.

위 urdf의 경우를 살펴보면

`<origin rpy="0 1.57075 0" xyz="0 0 -0.3"/>`

링크 안에 있는 이 origin의 경우에는 

직육면체 링크(`box`)를 자체 좌표계 기준으로

pitch를 $\frac{\pi}{2}$, 즉 90도 회전시키라는 의미이며,

추가로 링크 자체 좌표계를 기준으로 링크 모델을 z = -0.3 만큼 내리라는 의미이다.

`<origin xyz="0 -0.22 0.25"/>`

조인트 안에 있는 이 origin의 경우에는

`base_link` 좌표계를 기준으로 

x = 0, y = -0.22, z = 0.25 위치에 `child_link`의 좌표계를 배치하라는 의미이며,

자식 링크의 좌표계 자체가 움직였으니 링크도 당연히 따라 움직이는 결과를 낳는다.

이전 urdf처럼 만약 이 값을 주지 않는 경우에는 기본값 (0, 0, 0)에 배치된다. 

#### 04-materials.urdf 실습

실행

```bash
ros2 launch urdf_tutorial display.launch.py model:=urdf/04-materials.urdf
```

rviz창과 같이 뜨는 또 하나의 창에서 center를 눌러주고

이후 rviz에 모양이 나오는지 확인한다. 시간이 일부 소요된다.

내용확인

```bash
cat /opt/ros/humble/share/urdf_tutorial/urdf/04-materials.urdf
```

````python
<?xml version="1.0"?>
<robot name="materials">

  <material name="blue">
    <color rgba="0 0 0.8 1"/>
  </material>

  <material name="white">
    <color rgba="1 1 1 1"/>
  </material>


  <link name="base_link">
    <visual>
      <geometry>
        <cylinder length="0.6" radius="0.2"/>
      </geometry>
      <material name="blue"/>
    </visual>
  </link>

  <link name="right_leg">
    <visual>
      <geometry>
        <box size="0.6 0.1 0.2"/>
      </geometry>
      <origin rpy="0 1.57075 0" xyz="0 0 -0.3"/>
      <material name="white"/>
    </visual>
  </link>

  <joint name="base_to_right_leg" type="fixed">
    <parent link="base_link"/>
    <child link="right_leg"/>
    <origin xyz="0 -0.22 0.25"/>
  </joint>

  <link name="left_leg">
    <visual>
      <geometry>
        <box size="0.6 0.1 0.2"/>
      </geometry>
      <origin rpy="0 1.57075 0" xyz="0 0 -0.3"/>
      <material name="white"/>
    </visual>
  </link>

  <joint name="base_to_left_leg" type="fixed">
    <parent link="base_link"/>
    <child link="left_leg"/>
    <origin xyz="0 0.22 0.25"/>
  </joint>

</robot>
````

이번엔 `<material>`이 추가되었다.

물론 `left_leg`라는 링크도 추가되었지만 구조가 `right_leg`와 동일하므로 생략한다.

`<material>`은 재질을 정의하는 부분이다.

```xml
<material name="blue">
  <color rgba="0 0 0.8 1"/>
</material>
```

`blue`라는 이름의 재질을 정의하는 부분이다.

rgba로 색상을 정의하는데, 각각 red, green, blue, alpha(투명도)이다.

각각 비율을 나타내며, alpha의 경우 0이면 완전 투명(안보임), 1이면 완전 선명이다.

즉, 위 코드는 `blue`라는 재질은 파란색임을 나타내는 것이다.

`white` 재질 코드도 바로 아래에 나오는데, 구조가 동일하므로 생략한다.

링크 안쪽에 보면, `<material name="blue"/>`, `<material name="white"/>` 처럼

해당 링크가 어떤 재질을 사용하는지를 정의하는 부분이 있다.

### 움직이는 모델 만들기

실행

```bash
ros2 launch urdf_tutorial display.launch.py model:=urdf/06-flexible.urdf
```

이후 rviz에 로봇과 함께 별도의 스크롤 창이 나오는지 확인한다. 시간이 일부 소요된다.

![flexible.urdf](../imgs/urdf1.png)

확인되었으면

스크롤 창의 각 요소들을 스크롤해보면서 회전 또는 직선이동이 잘 되는지 확인한다.

내용확인

```bash
cat /opt/ros/humble/share/urdf_tutorial/urdf/06-flexible.urdf
```

````python
<joint name="head_swivel" type="continuous">
    <parent link="base_link"/>
    <child link="head"/>
    <axis xyz="0 0 1"/>
    <origin xyz="0 0 0.3"/>
</joint>
````

이전 코드에서는 joint의 type이 전부 `fixed`로 고정이었지만

머리 링크의 경우 `continous`, 즉 연속 조인트로 모델링되었다.

연속 조인트는 모든 각도를 취할 수 있다.

````python
<joint name="left_gripper_joint" type="revolute">
    <axis xyz="0 0 1"/>
    <limit effort="1000.0" lower="0.0" upper="0.548" velocity="0.5"/>
    <origin rpy="0 0 0" xyz="0.2 0.01 0"/>
    <parent link="gripper_pole"/>
    <child link="left_gripper"/>
</joint>
````

그리퍼 링크의 경우 `revolute`, 즉 회전 조인트로 모델링되었다.

아래 코드는 회전 조인트의 토크(effort), 하한(lower), 상한(upper), 최대속도(velocity) 등을 정의한다.

```xml
<limit effort="1000.0" lower="0.0" upper="0.548" velocity="0.5"/>
```

`effort` : 조인트가 낼 수 있는 최대 토크 한도(N $\cdot$ m. 설명을 위한 예시일 뿐 실제 이 단위는 아님.)

`lower`, `upper` : 하한, 상한각도. 즉 0 - 0.548rad까지만 회전가능.

`velocity` : 최대 속도(rad/s)

````python
<joint name="gripper_extension" type="prismatic">
    <parent link="base_link"/>
    <child link="gripper_pole"/>
    <limit effort="1000.0" lower="-0.38" upper="0" velocity="0.5"/>
    <origin rpy="0 0 0" xyz="0.19 0 0.2"/>
</joint>
````

그리퍼 암 링크의 경우 `prismatic`, 즉 직선 운동 조인트로 모델링되었다.

회전 조인트와 동일하게 `effort`, `lower`, `upper`, `velocity`로 제한을 정의하는데,

회전이 아닌 직선 운동이기 때문에 제약의 의미가 살짝 달라진다.

`effort` : 조인트가 낼 수 있는 최대 힘 한도(N. 설명을 위한 예시일 뿐 실제 이 단위는 아님.)

`lower`, `upper` : 하한, 상한위치. 즉 -0.38 ~ 0까지만 이동가능.

`velocity` : 최대 속도

#### 충돌, 관성 속성 추가

내용확인

```bash
cat /opt/ros/humble/share/urdf_tutorial/urdf/07-physics.urdf
```

````python
<collision>
  <geometry>
    <cylinder length="0.6" radius="0.2"/>
  </geometry>
</collision>
````

기존의 `base_link` 안의 내용에 위와 같은 `<collision>` 부분이 추가되었다.

`<collision>`은 `충돌 판정용 모양`으로, 충돌 범위라고 생각하면 된다.

즉 두 물체의 충돌영역이 만나거나 겹치면 충돌 발생으로 판단하는 것이다.

위의 경우 `base_link`의 원통형 모양과 완벽하게 똑같이 `<collision>`을 만들었으므로

원통형 모양 전체를 충돌 영역으로 설정한 것이다.

````python
<inertial>
    <mass value="10"/>
    <inertia ixx="1e-3" ixy="0.0" ixz="0.0" iyy="1e-3" iyz="0.0" izz="1e-3"/>
</inertial>
````

관성 속성

`mass` : 질량

`ixx`, `iyy`, `izz` : x, y, z축을 중심으로 하는 회전저항(해당 축을 기준으로 얼마나 돌리기 어려운가)

`ixy`, `ixz`, `iyz` : 관성곱(한 축으로 돌렸는데 다른 축 회전까지 같이 발생하는 정도)
